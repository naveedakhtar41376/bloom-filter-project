# Bloom Filter Analysis Notebook

This notebook is for demonstrating the implementation and summarising the experiment outputs. Run the scripts first to generate the CSV files in `results/`.


## 1. Basic usage


```python
from bloomfilter import BloomFilter

bf = BloomFilter(expected_items=1000, false_positive_rate=0.01)
bf.add('apple')

print('apple' in bf)
print('banana' in bf)
print(bf.config)
```

## 2. Load experiment results


```python
from pathlib import Path
import pandas as pd

results_dir = Path('../results')
benchmark = pd.read_csv(results_dir / 'benchmark_results.csv')
false_positive = pd.read_csv(results_dir / 'false_positive_results.csv')
compression = pd.read_csv(results_dir / 'compression_results.csv')
hash_dist = pd.read_csv(results_dir / 'hash_distribution_results.csv')

benchmark.head()
```

## 3. Benchmark results

Insertion and search are expected to scale approximately linearly with the number of operations because each operation uses a fixed number of hash functions for a configured filter.


```python
benchmark[['data_type', 'n', 'insert_time_seconds', 'positive_search_time_seconds', 'negative_search_time_seconds', 'num_hashes', 'memory_bytes']]
```

## 4. False-positive-rate behaviour

The empirical false-positive rate should be close to the theoretical value near the designed capacity. Once inserted items exceed the design capacity, the false-positive rate should increase.


```python
false_positive[['capacity_ratio', 'empirical_fpr', 'theoretical_fpr', 'fill_ratio']]
```
